```python
from tensorflow.keras.datasets import fashion_mnist

# Load dataset
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()
```

```python
# normalize the data
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten

model = Sequential(
    [
        Flatten(input_shape=(28, 28)),
        Dense(128, activation='sigmoid', input_shape=(28 * 28,)),
        Dense(64, activation='sigmoid'),
        Dense(32, activation='sigmoid'),
        Dense(10, activation='softmax')
    ]
)

model.compile(
    optimizer='sgd',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(X_train, y_train, epochs=10)
```

```python
model.evaluate(X_test, y_test)
```

```python
# make new convolutional model
from tensorflow.keras.layers import Conv2D

model2 = Sequential(
    [
        Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(28, 28, 1)),
        Conv2D(64, kernel_size=(3, 3), activation='relu'),
        Flatten(),
        Dense(128, activation='relu'),
        Dense(10, activation='softmax')
    ]
)

model2.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model2.fit(X_train.reshape(-1, 28, 28, 1), y_train, epochs=2)
```

```python
model2.evaluate(X_test.reshape(-1, 28, 28, 1), y_test)
```

```python
from tensorflow.keras.layers import MaxPooling2D

model3 = Sequential([
    Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D(pool_size=(2, 2)),
    
    Conv2D(64, kernel_size=(3, 3), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),
    
    Flatten(),    

    Dense(128, activation='sigmoid'),
    Dense(10, activation='softmax')
])

model3.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model3.fit(X_train.reshape(-1, 28, 28, 1), y_train, epochs=3)
```

```python
model3.evaluate(X_test.reshape(-1, 28, 28, 1), y_test)
```

```python
# AlexNet time

from tensorflow.keras.layers import Input, BatchNormalization, Dropout


model4 = Sequential(
    [
        Input(shape=(28, 28, 1)),
        Conv2D(32, kernel_size=(3, 3), activation="relu", padding="same"),
        BatchNormalization(),
        MaxPooling2D(pool_size=(2, 2)),
        Conv2D(64, kernel_size=(3, 3), activation="relu", padding="same"),
        BatchNormalization(),
        MaxPooling2D(pool_size=(2, 2)),
        Conv2D(128, kernel_size=(3, 3), activation="relu", padding="same"),
        Conv2D(128, kernel_size=(3, 3), activation="relu", padding="same"),
        Conv2D(64, kernel_size=(3, 3), activation="relu", padding="same"),
        MaxPooling2D(pool_size=(2, 2)),
        Flatten(),
        Dense(128, activation="relu"),
        Dropout(0.5),
        Dense(128, activation="relu"),
        Dropout(0.5),
        Dense(10, activation="softmax"),
    ]
)

model4.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# graph training loss and accuracy using tensorboard
from tensorflow.keras.callbacks import TensorBoard

tensorboard_callback = TensorBoard(log_dir='./logs', histogram_freq=1)
model4.fit(X_train.reshape(-1, 28, 28, 1), y_train, epochs=10, callbacks=[tensorboard_callback])
```

# Conclusion
After experimenting with several neural network architectures on the Fashion MNIST dataset, we observed the following:

- Simple dense (fully connected) networks can achieve reasonable accuracy, but their performance is limited compared to convolutional models.
- Introducing convolutional layers (model2) significantly improved the model's ability to extract spatial features, resulting in higher accuracy.
- However, adding max pooling layers (model3) and using a more complex AlexNet-inspired architecture (model4) slightly diminished the accuracy, possibly because the models became too detailed or overfit to the training data.
- This suggests that, for this dataset, a moderate convolutional architecture without excessive complexity or pooling may yield the best results.

In summary, convolutional neural networks outperform simple dense networks on image data like Fashion MNIST, but increasing architectural complexity does not always guarantee better performance. Future improvements could include data augmentation, regularization tuning, or exploring other architectures suited to the dataset's characteristics.
